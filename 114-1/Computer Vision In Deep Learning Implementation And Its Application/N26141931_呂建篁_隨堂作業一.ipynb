```python
#作業第一題

import matplotlib.pyplot as plt

# 定義點
x = [0, 4, 6]
y = [0, 4, 0]

# 畫折線
plt.plot(x, y, color='orange', linewidth=4)

# 設定座標軸範圍
plt.xlim(-4, 13)
plt.ylim(-5, 7)

# 畫 x=0 和 y=0 的坐標軸
plt.axhline(0, color='black')
plt.axvline(0, color='black')

# 加上 "k" 標籤在 x 軸上接近 0 的地方
plt.text(0, -0.5, 'k', color='orange')

# 格線
plt.grid(True)

plt.gca().set_aspect('equal', adjustable='box')  # 保持方格比例一致
plt.show()
```

```python
#作業二

import numpy as np
import matplotlib.pyplot as plt

# 左半圓 (向下)
x1 = np.linspace(0, 4, 200)
y1 = -np.sqrt(4 - (x1 - 2)**2)

# 水平線
x2 = np.linspace(4, 6, 50)
y2 = np.zeros_like(x2)

# 右半圓 (向上)
x3 = np.linspace(6, 10, 200)
y3 = np.sqrt(4 - (x3 - 8)**2)

# 繪圖
plt.figure(figsize=(8,6))
plt.plot(x1, y1, color="lime", linewidth=3)
plt.plot(x2, y2, color="lime", linewidth=3)
plt.plot(x3, y3, color="lime", linewidth=3)

# 座標軸
plt.axhline(0, color="black")
plt.axvline(0, color="black")

# 格線
plt.grid(True, which='both', linestyle='--', linewidth=0.5)

plt.show()
```

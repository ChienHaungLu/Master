```python
import json
import os
from pathlib import Path

def manual_convert_coco_to_yolo(json_file, save_dir):
    # 1. 檢查 JSON 檔案是否存在
    json_path = Path(json_file)
    if not json_path.exists():
        print(f"錯誤：找不到檔案 {json_path}")
        return

    # 2. 建立儲存資料夾
    save_path = Path(save_dir)
    save_path.mkdir(parents=True, exist_ok=True)
    print(f"輸出目錄設定為: {save_path}")

    # 3. 讀取 JSON
    print("正在讀取 JSON 檔...")
    with open(json_path, 'r', encoding='utf-8') as f:
        data = json.load(f)

    # 建立圖片 ID 對應資訊的字典 (用來查圖片長寬)
    # 結構: {image_id: {'file_name':Str, 'width':Int, 'height':Int}}
    images_info = {}
    for img in data['images']:
        images_info[img['id']] = {
            'file_name': img['file_name'],
            'width': img['width'],
            'height': img['height']
        }

    # 4. 開始轉換標註
    print(f"找到 {len(data['annotations'])} 個標註，開始轉換...")
    
    count = 0
    # 用來暫存要寫入每個檔案的內容: {filename: [line1, line2...]}
    file_contents = {}

    for ann in data['annotations']:
        image_id = ann['image_id']
        category_id = ann['category_id'] # 注意：如果你的類別是從1開始，YOLO需要從0開始，可能需要 -1
        bbox = ann['bbox'] # COCO 格式: [x_min, y_min, width, height]

        if image_id not in images_info:
            continue

        img_w = images_info[image_id]['width']
        img_h = images_info[image_id]['height']
        file_name = images_info[image_id]['file_name']
        
        # 轉換檔名：把 .jpg/.png 換成 .txt
        txt_name = os.path.splitext(os.path.basename(file_name))[0] + ".txt"

        # --- 核心數學轉換 ---
        x_min, y_min, w, h = bbox
        
        # 計算中心點
        x_center = x_min + (w / 2)
        y_center = y_min + (h / 2)

        # 歸一化 (Normalize) 到 0~1 之間
        x_c_norm = x_center / img_w
        y_c_norm = y_center / img_h
        w_norm = w / img_w
        h_norm = h / img_h
        
        # 限制數值在 0-1 之間 (防止標註超出圖片邊界)
        x_c_norm = max(0.0, min(1.0, x_c_norm))
        y_c_norm = max(0.0, min(1.0, y_c_norm))
        w_norm = max(0.0, min(1.0, w_norm))
        h_norm = max(0.0, min(1.0, h_norm))

        # 格式化 YOLO 字串: class x_center y_center width height
        # 假設你的 category_id 不需要調整 (如果你的 dataset class id 是從 1 開始，這裡要改成 category_id - 1)
        yolo_line = f"{category_id-1} {x_c_norm:.6f} {y_c_norm:.6f} {w_norm:.6f} {h_norm:.6f}"

        if txt_name not in file_contents:
            file_contents[txt_name] = []
        file_contents[txt_name].append(yolo_line)
        count += 1

    # 5. 寫入檔案
    for txt_name, lines in file_contents.items():
        out_file = save_path / txt_name
        with open(out_file, 'w') as f:
            f.write('\n'.join(lines))

    print(f"轉換成功！")
    print(f"- 處理了 {len(file_contents)} 張圖片的標註")
    print(f"- 總共生成 {count} 個 bounding box")
    print(f"- 檔案儲存在: {save_path}")

# --- 設定路徑並執行 ---
count = 0
for jsonf in["../raw/positive-20251022T114300Z-1-001/positive/annotations.json","../raw/positive-20251022T114300Z-1-002/positive/instances_default.json","../raw/positive-20251022T114300Z-1-003/positive/instances_default.json"]:
    print(f"Processing {jsonf}...")
    yolo_folder = os.path.join(DATASET_FOLDER, f"{count}")
    count += 1
    manual_convert_coco_to_yolo(jsonf, yolo_folder)


```

```python
# ...existing code...
import os
import shutil
from pathlib import Path

# Uses DATASET_FOLDER defined earlier in this notebook
src_root = Path(DATASET_FOLDER)
dst = src_root / "merged"
dst.mkdir(parents=True, exist_ok=True)

# Extensions to collect
exts = {".png", ".jpg", ".jpeg", ".PNG", ".JPG", ".JPEG", ".txt"}

copied = 0
for sub in sorted(p for p in src_root.iterdir() if p.is_dir()):
    prefix = sub.name
    for f in sub.iterdir():
        if f.is_file() and f.suffix in exts:
            # prefix to avoid name collisions across subfolders
            new_name = f"{prefix}__{f.name}"
            dest_path = dst / new_name

            # if still exists, add numeric suffix
            if dest_path.exists():
                base = dest_path.stem
                suf = dest_path.suffix
                i = 1
                while (dst / f"{base}_{i}{suf}").exists():
                    i += 1
                dest_path = dst / f"{base}_{i}{suf}"

            shutil.copy2(f, dest_path)
            copied += 1

print(f"Done. Copied {copied} files into: {dst}")
# ...existing code...
```

```python
# ...existing code...
from pathlib import Path
from collections import Counter, defaultdict
import json, csv, statistics, math
try:
    import matplotlib.pyplot as plt
    import seaborn as sns
except Exception:
    plt = None
    sns = None

ROOT = Path(DATASET_FOLDER)
MERGED = ROOT / "merged"
PLOTS = ROOT / "plots"
PLOTS.mkdir(parents=True, exist_ok=True)
OUT_JSON = ROOT / "report_stats.json"
OUT_CSV = ROOT / "report_per_image.csv"

# 1) load image sizes (try png_stats.json then scan images)
img_sizes = {}  # stem -> (w,h)
png_stats = ROOT / "png_stats.json"
if png_stats.exists():
    try:
        with open(png_stats, "r") as f:
            d = json.load(f)
        rows = d.get("rows") or d.get("images") or []
        for r in rows:
            stem = r.get("stem") or Path(r.get("path","")).stem
            if stem:
                img_sizes[stem] = (r.get("width"), r.get("height"))
    except Exception:
        img_sizes = {}

if not img_sizes:
    try:
        from PIL import Image
        img_exts = {".png", ".jpg", ".jpeg", ".bmp", ".tif", ".tiff"}
        for p in MERGED.iterdir():
            if p.is_file() and p.suffix.lower() in img_exts:
                try:
                    with Image.open(p) as im:
                        img_sizes[p.stem] = im.size  # (w,h)
                except Exception:
                    pass
    except Exception:
        img_sizes = {}

# 2) list label files in merged and parse
label_files = sorted([p for p in MERGED.iterdir() if p.is_file() and p.suffix.lower()==".txt"])
per_image = {}   # stem -> info
class_counts = Counter()
bbox_areas_norm = []   # normalized w*h across dataset
bbox_widths_norm = []
bbox_heights_norm = []

for lf in label_files:
    stem = lf.stem
    with open(lf, "r") as f:
        lines = [L.strip() for L in f if L.strip()]
    boxes = []
    for L in lines:
        parts = L.split()
        if len(parts) >= 5:
            try:
                cls = int(float(parts[0]))
                x = float(parts[1]); y = float(parts[2])
                w = float(parts[3]); h = float(parts[4])
                area = max(0.0, w*h)
                boxes.append({"class":cls,"x":x,"y":y,"w":w,"h":h,"area_norm":area})
                class_counts[cls] += 1
                bbox_areas_norm.append(area)
                bbox_widths_norm.append(w)
                bbox_heights_norm.append(h)
            except Exception:
                continue
        elif parts:
            try:
                cls = int(float(parts[0])); class_counts[cls]+=1
            except Exception:
                pass
    # pixel areas if image size known
    img_area = None
    if stem in img_sizes and img_sizes[stem][0] and img_sizes[stem][1]:
        iw, ih = img_sizes[stem]
        img_area = iw * ih
    total_obj_area_norm = sum(b["area_norm"] for b in boxes)
    total_obj_area_pixels = None
    if img_area:
        total_obj_area_pixels = total_obj_area_norm * img_area
    per_image[stem] = {
        "num_boxes": len(boxes),
        "boxes": boxes,
        "img_size": img_sizes.get(stem),
        "total_obj_area_norm": total_obj_area_norm,
        "total_obj_area_pixels": total_obj_area_pixels,
        "coverage_pct": (total_obj_area_norm * 100.0) if img_area else None
    }

# 3) read autosplit if available (train/val/test filenames)
splits = {}
for name in ("autosplit_train.txt","autosplit_val.txt","autosplit_test.txt"):
    p = ROOT / name
    if p.exists():
        with open(p,"r") as f:
            items = [Path(L.strip()).stem for L in f if L.strip()]
        splits[name.split("_")[1].split(".")[0]] = items
if not splits:
    splits["all"] = list(per_image.keys())

# 4) compute summary stats
def stats(xs):
    if not xs: return {}
    return {
        "count": len(xs),
        "min": min(xs),
        "max": max(xs),
        "mean": statistics.mean(xs),
        "median": statistics.median(xs),
        "stdev": statistics.pstdev(xs) if len(xs)>1 else 0.0
    }

num_images = len(list(MERGED.glob("*")))-len([p for p in MERGED.glob("*.txt")])  # rough images count
num_label_files = len(label_files)
total_bboxes = sum(v["num_boxes"] for v in per_image.values())
images_with_labels = len(per_image)
images_without_label = max(0, num_images - images_with_labels) if num_images>=0 else 0

summary = {
    "num_files_in_merged": len(list(MERGED.iterdir())),
    "num_images_estimate": num_images,
    "num_label_files": num_label_files,
    "images_with_label": images_with_labels,
    "images_without_label_estimate": images_without_label,
    "total_bboxes": total_bboxes,
    "avg_boxes_per_labeled_image": statistics.mean([v["num_boxes"] for v in per_image.values()]) if per_image else 0,
    "class_counts": dict(class_counts),
    "bbox_area_norm_stats": stats(bbox_areas_norm),
    "bbox_width_norm_stats": stats(bbox_widths_norm),
    "bbox_height_norm_stats": stats(bbox_heights_norm),
    "per_image_coverage_stats_pct": stats([v["coverage_pct"] for v in per_image.values() if v["coverage_pct"] is not None])
}

# 5) per-split summaries
split_summaries = {}
for split_name, stems in splits.items():
    s_per_image = [per_image[s] for s in stems if s in per_image]
    s_total_boxes = sum(x["num_boxes"] for x in s_per_image)
    s_coverages = [x["coverage_pct"] for x in s_per_image if x["coverage_pct"] is not None]
    split_summaries[split_name] = {
        "n_images_in_split_file": len(stems),
        "n_labeled_images_in_split": len(s_per_image),
        "total_boxes": s_total_boxes,
        "avg_boxes_per_labeled_image": statistics.mean([x["num_boxes"] for x in s_per_image]) if s_per_image else 0,
        "coverage_pct_stats": stats(s_coverages)
    }
summary["splits"] = split_summaries

# 6) save per-image table CSV
with open(OUT_CSV, "w", newline="") as f:
    w = csv.writer(f)
    w.writerow(["stem","num_boxes","img_width","img_height","total_obj_area_norm","coverage_pct"])
    for stem, info in per_image.items():
        w.writerow([stem, info["num_boxes"],
                    info["img_size"][0] if info["img_size"] else None,
                    info["img_size"][1] if info["img_size"] else None,
                    round(info["total_obj_area_norm"],6),
                    round(info["coverage_pct"],4) if info["coverage_pct"] is not None else None])

# 7) save summary JSON
with open(OUT_JSON, "w") as f:
    json.dump({"summary": summary, "n_images_with_label": images_with_labels}, f, indent=2)

# 8) optional plots (if matplotlib available) -> save to plots folder
if plt and sns:
    sns.set(style="whitegrid")
    # class distribution bar
    fig, ax = plt.subplots(figsize=(6,4))
    items = sorted(summary["class_counts"].items())
    if items:
        classes, counts = zip(*items)
        sns.barplot(x=list(classes), y=list(counts), ax=ax, palette="Blues_d")
        ax.set_xlabel("class"); ax.set_ylabel("count"); ax.set_title("Class distribution")
        fig.savefig(PLOTS / "report_class_counts.png", bbox_inches="tight", dpi=150); plt.close(fig)
    # boxes per image histogram
    fig, ax = plt.subplots(figsize=(6,4))
    vals = [v["num_boxes"] for v in per_image.values()]
    sns.histplot(vals, bins=min(50, max(5, int(math.sqrt(len(vals))))), ax=ax)
    ax.set_xlabel("boxes per labeled image"); ax.set_title("Boxes per labeled image")
    fig.savefig(PLOTS / "report_boxes_per_image.png", bbox_inches="tight", dpi=150); plt.close(fig)
    # coverage CDF
    coverages = [v["coverage_pct"] for v in per_image.values() if v["coverage_pct"] is not None]
    if coverages:
        fig, ax = plt.subplots(figsize=(6,4))
        sns.ecdfplot(coverages, ax=ax)
        ax.set_xlabel("total object coverage (%)"); ax.set_title("CDF of object area coverage per image")
        fig.savefig(PLOTS / "report_coverage_cdf.png", bbox_inches="tight", dpi=150); plt.close(fig)

print("Report stats saved:", OUT_JSON)
print("Per-image CSV saved:", OUT_CSV)
print("Plots saved to:", PLOTS)
# ...existing code...
# filepath: /media/james/james/Course/ML/Final_Project/NCKU-ML-25Fall/Final_Project/ipynb/yolo.ipynb
# ...existing code...
from pathlib import Path
from collections import Counter, defaultdict
import json, csv, statistics, math
try:
    import matplotlib.pyplot as plt
    import seaborn as sns
except Exception:
    plt = None
    sns = None

ROOT = Path(DATASET_FOLDER)
MERGED = ROOT / "merged"
PLOTS = ROOT / "plots"
PLOTS.mkdir(parents=True, exist_ok=True)
OUT_JSON = ROOT / "report_stats.json"
OUT_CSV = ROOT / "report_per_image.csv"

# 1) load image sizes (try png_stats.json then scan images)
img_sizes = {}  # stem -> (w,h)
png_stats = ROOT / "png_stats.json"
if png_stats.exists():
    try:
        with open(png_stats, "r") as f:
            d = json.load(f)
        rows = d.get("rows") or d.get("images") or []
        for r in rows:
            stem = r.get("stem") or Path(r.get("path","")).stem
            if stem:
                img_sizes[stem] = (r.get("width"), r.get("height"))
    except Exception:
        img_sizes = {}

if not img_sizes:
    try:
        from PIL import Image
        img_exts = {".png", ".jpg", ".jpeg", ".bmp", ".tif", ".tiff"}
        for p in MERGED.iterdir():
            if p.is_file() and p.suffix.lower() in img_exts:
                try:
                    with Image.open(p) as im:
                        img_sizes[p.stem] = im.size  # (w,h)
                except Exception:
                    pass
    except Exception:
        img_sizes = {}

# 2) list label files in merged and parse
label_files = sorted([p for p in MERGED.iterdir() if p.is_file() and p.suffix.lower()==".txt"])
per_image = {}   # stem -> info
class_counts = Counter()
bbox_areas_norm = []   # normalized w*h across dataset
bbox_widths_norm = []
bbox_heights_norm = []

for lf in label_files:
    stem = lf.stem
    with open(lf, "r") as f:
        lines = [L.strip() for L in f if L.strip()]
    boxes = []
    for L in lines:
        parts = L.split()
        if len(parts) >= 5:
            try:
                cls = int(float(parts[0]))
                x = float(parts[1]); y = float(parts[2])
                w = float(parts[3]); h = float(parts[4])
                area = max(0.0, w*h)
                boxes.append({"class":cls,"x":x,"y":y,"w":w,"h":h,"area_norm":area})
                class_counts[cls] += 1
                bbox_areas_norm.append(area)
                bbox_widths_norm.append(w)
                bbox_heights_norm.append(h)
            except Exception:
                continue
        elif parts:
            try:
                cls = int(float(parts[0])); class_counts[cls]+=1
            except Exception:
                pass
    # pixel areas if image size known
    img_area = None
    if stem in img_sizes and img_sizes[stem][0] and img_sizes[stem][1]:
        iw, ih = img_sizes[stem]
        img_area = iw * ih
    total_obj_area_norm = sum(b["area_norm"] for b in boxes)
    total_obj_area_pixels = None
    if img_area:
        total_obj_area_pixels = total_obj_area_norm * img_area
    per_image[stem] = {
        "num_boxes": len(boxes),
        "boxes": boxes,
        "img_size": img_sizes.get(stem),
        "total_obj_area_norm": total_obj_area_norm,
        "total_obj_area_pixels": total_obj_area_pixels,
        "coverage_pct": (total_obj_area_norm * 100.0) if img_area else None
    }

# 3) read autosplit if available (train/val/test filenames)
splits = {}
for name in ("autosplit_train.txt","autosplit_val.txt","autosplit_test.txt"):
    p = ROOT / name
    if p.exists():
        with open(p,"r") as f:
            items = [Path(L.strip()).stem for L in f if L.strip()]
        splits[name.split("_")[1].split(".")[0]] = items
if not splits:
    splits["all"] = list(per_image.keys())

# 4) compute summary stats
def stats(xs):
    if not xs: return {}
    return {
        "count": len(xs),
        "min": min(xs),
        "max": max(xs),
        "mean": statistics.mean(xs),
        "median": statistics.median(xs),
        "stdev": statistics.pstdev(xs) if len(xs)>1 else 0.0
    }

num_images = len(list(MERGED.glob("*")))-len([p for p in MERGED.glob("*.txt")])  # rough images count
num_label_files = len(label_files)
total_bboxes = sum(v["num_boxes"] for v in per_image.values())
images_with_labels = len(per_image)
images_without_label = max(0, num_images - images_with_labels) if num_images>=0 else 0

summary = {
    "num_files_in_merged": len(list(MERGED.iterdir())),
    "num_images_estimate": num_images,
    "num_label_files": num_label_files,
    "images_with_label": images_with_labels,
    "images_without_label_estimate": images_without_label,
    "total_bboxes": total_bboxes,
    "avg_boxes_per_labeled_image": statistics.mean([v["num_boxes"] for v in per_image.values()]) if per_image else 0,
    "class_counts": dict(class_counts),
    "bbox_area_norm_stats": stats(bbox_areas_norm),
    "bbox_width_norm_stats": stats(bbox_widths_norm),
    "bbox_height_norm_stats": stats(bbox_heights_norm),
    "per_image_coverage_stats_pct": stats([v["coverage_pct"] for v in per_image.values() if v["coverage_pct"] is not None])
}

# 5) per-split summaries
split_summaries = {}
for split_name, stems in splits.items():
    s_per_image = [per_image[s] for s in stems if s in per_image]
    s_total_boxes = sum(x["num_boxes"] for x in s_per_image)
    s_coverages = [x["coverage_pct"] for x in s_per_image if x["coverage_pct"] is not None]
    split_summaries[split_name] = {
        "n_images_in_split_file": len(stems),
        "n_labeled_images_in_split": len(s_per_image),
        "total_boxes": s_total_boxes,
        "avg_boxes_per_labeled_image": statistics.mean([x["num_boxes"] for x in s_per_image]) if s_per_image else 0,
        "coverage_pct_stats": stats(s_coverages)
    }
summary["splits"] = split_summaries

# 6) save per-image table CSV
with open(OUT_CSV, "w", newline="") as f:
    w = csv.writer(f)
    w.writerow(["stem","num_boxes","img_width","img_height","total_obj_area_norm","coverage_pct"])
    for stem, info in per_image.items():
        w.writerow([stem, info["num_boxes"],
                    info["img_size"][0] if info["img_size"] else None,
                    info["img_size"][1] if info["img_size"] else None,
                    round(info["total_obj_area_norm"],6),
                    round(info["coverage_pct"],4) if info["coverage_pct"] is not None else None])

# 7) save summary JSON
with open(OUT_JSON, "w") as f:
    json.dump({"summary": summary, "n_images_with_label": images_with_labels}, f, indent=2)

# 8) optional plots (if matplotlib available) -> save to plots folder
if plt and sns:
    sns.set(style="whitegrid")
    # class distribution bar
    fig, ax = plt.subplots(figsize=(6,4))
    items = sorted(summary["class_counts"].items())
    if items:
        classes, counts = zip(*items)
        sns.barplot(x=list(classes), y=list(counts), ax=ax, palette="Blues_d")
        ax.set_xlabel("class"); ax.set_ylabel("count"); ax.set_title("Class distribution")
        fig.savefig(PLOTS / "report_class_counts.png", bbox_inches="tight", dpi=150); plt.close(fig)
    # boxes per image histogram
    fig, ax = plt.subplots(figsize=(6,4))
    vals = [v["num_boxes"] for v in per_image.values()]
    sns.histplot(vals, bins=min(50, max(5, int(math.sqrt(len(vals))))), ax=ax)
    ax.set_xlabel("boxes per labeled image"); ax.set_title("Boxes per labeled image")
    fig.savefig(PLOTS / "report_boxes_per_image.png", bbox_inches="tight", dpi=150); plt.close(fig)
    # coverage CDF
    coverages = [v["coverage_pct"] for v in per_image.values() if v["coverage_pct"] is not None]
    if coverages:
        fig, ax = plt.subplots(figsize=(6,4))
        sns.ecdfplot(coverages, ax=ax)
        ax.set_xlabel("total object coverage (%)"); ax.set_title("CDF of object area coverage per image")
        fig.savefig(PLOTS / "report_coverage_cdf.png", bbox_inches="tight", dpi=150); plt.close(fig)

print("Report stats saved:", OUT_JSON)
print("Per-image CSV saved:", OUT_CSV)
print("Plots saved to:", PLOTS)
# ...existing code...
```

```python
# ...existing code...
from pathlib import Path
from collections import Counter

# 使用 notebook 中定義的 DATASET_FOLDER（若在 script 執行，可改為直接指定路徑）
ROOT = Path(DATASET_FOLDER)  # see: [`DATASET_FOLDER`](Final_Project/ipynb/yolo.ipynb)
IMAGES_DIR = ROOT / "images"  # see: [Final_Project/dataset_final/images](Final_Project/dataset_final/images)
LABELS_DIR = ROOT / "labels"  # see: [Final_Project/dataset_final/labels](Final_Project/dataset_final/labels)

img_exts = {".png", ".jpg", ".jpeg", ".bmp", ".tif", ".tiff"}
splits = ["train", "test"]

summary = {}
total_images = 0
total_objects = 0
per_class = Counter()

for s in splits:
    imgs = []
    lbls = []
    if (IMAGES_DIR / s).exists():
        imgs = [p for p in (IMAGES_DIR / s).iterdir() if p.is_file() and p.suffix.lower() in img_exts]
    if (LABELS_DIR / s).exists():
        lbls = sorted(p for p in (LABELS_DIR / s).iterdir() if p.is_file() and p.suffix.lower() == ".txt")
    # image count
    n_imgs = len(imgs)
    # object count from label files (每一行視為一個物件)
    n_objs = 0
    for lf in lbls:
        try:
            lines = [L for L in lf.read_text(encoding="utf-8").splitlines() if L.strip() and not L.strip().startswith("#")]
            n_objs += len(lines)
            for L in lines:
                parts = L.split()
                if parts:
                    try:
                        cls = int(float(parts[0]))
                        per_class[cls] += 1
                    except Exception:
                        continue
        except Exception as e:
            print(f"Warning: failed to read {lf}: {e}")
    summary[s] = {"n_images": n_imgs, "n_label_files": len(lbls), "n_objects": n_objs}
    total_images += n_imgs
    total_objects += n_objs

# also count any images/labels at ROOT level (if applicable)
root_imgs = [p for p in ROOT.iterdir() if p.is_file() and p.suffix.lower() in img_exts]
root_lbls = [p for p in ROOT.iterdir() if p.is_file() and p.suffix.lower() == ".txt"]
if root_imgs or root_lbls:
    n_root_imgs = len(root_imgs)
    n_root_objs = 0
    for lf in root_lbls:
        try:
            lines = [L for L in lf.read_text(encoding="utf-8").splitlines() if L.strip() and not L.strip().startswith("#")]
            n_root_objs += len(lines)
            for L in lines:
                parts = L.split()
                if parts:
                    try:
                        cls = int(float(parts[0]))
                        per_class[cls] += 1
                    except Exception:
                        continue
        except Exception as e:
            print(f"Warning: failed to read {lf}: {e}")
    summary["root"] = {"n_images": n_root_imgs, "n_label_files": len(root_lbls), "n_objects": n_root_objs}
    total_images += n_root_imgs
    total_objects += n_root_objs

print("Dataset root:", ROOT)  # see: [Final_Project/dataset_final](Final_Project/dataset_final)
print("Summary per split:", summary)
print("Total images:", total_images)
print("Total labelled objects:", total_objects)
print("Per-class object counts:", dict(sorted(per_class.items())))
# ...existing code...
```

```python
# ...existing code...
from pathlib import Path
from collections import Counter
import json, csv

ROOT = Path(DATASET_FOLDER)
MERGED = ROOT

txts = sorted(p for p in MERGED.iterdir() if p.is_file() and p.suffix.lower() == ".txt")
counter = Counter()
total_boxes = 0
for t in txts:
    try:
        for line in t.read_text(encoding="utf-8").splitlines():
            parts = line.strip().split()
            if not parts: 
                continue
            try:
                cls = int(float(parts[0]))
                counter[cls] += 1
                total_boxes += 1
            except Exception:
                continue
    except Exception as e:
        print(f"Warning: failed to read {t}: {e}")

out_json = ROOT / "class_counts.json"
out_csv = ROOT / "class_counts.csv"

# save json
with open(out_json, "w") as f:
    json.dump({"total_boxes": total_boxes, "class_counts": dict(counter)}, f, indent=2)

# save csv
with open(out_csv, "w", newline="") as f:
    w = csv.writer(f)
    w.writerow(["class_id","count"])
    for cls, cnt in sorted(counter.items()):
        w.writerow([cls, cnt])

print("Total labelled boxes:", total_boxes)
print("Classes:", sorted(counter.keys()))
print("Counts per class:", dict(counter))
print("Saved:", out_json, out_csv)
# ...existing code...
```

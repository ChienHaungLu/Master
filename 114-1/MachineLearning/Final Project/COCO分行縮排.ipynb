# Cell 1：上傳 JSON 檔案（支援 Colab）

```python
from google.colab import files

print("請上傳你的 COCO 1.0 JSON 檔案：") #不是COCO keypoint，是COCO 1.0
uploaded = files.upload()
```

# Cell 2 — 讀取並格式化 JSON（分行、縮排）

```python
import json

# 取得上傳的檔案名稱
filename = list(uploaded.keys())[0]

# 讀取 JSON
with open(filename, "r", encoding="utf-8") as f:
    coco_data = json.load(f)

# 格式化（分行+縮排）
pretty_json = json.dumps(coco_data, indent=4, ensure_ascii=False)

# 顯示格式化結果
print(pretty_json)
```

# Cell 3 — 輸出成新的 JSON 檔案

```python
output_name = "annotations.json" #預設annotations

with open(output_name, "w", encoding="utf-8") as f:
    f.write(pretty_json)

print(f"格式化後已輸出為：{output_name}")
```

# Cell 4 — 下載格式化後的 JSON

```python
from google.colab import files

files.download(output_name)
```

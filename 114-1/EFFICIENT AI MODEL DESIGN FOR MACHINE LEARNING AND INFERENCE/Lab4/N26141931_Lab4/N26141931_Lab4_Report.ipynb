## 2. Response-Based KD (12%)
> Please explain the following:
> - How you choose the Temperature and alpha?
> - How you design the loss?



*   在**Temperature**設定4且alpha設定0.5，在溫度的設定上主要的原因在於讓teacher logits較平滑來讓student學習類別之間的相似程度。在**alpha**設定的原因在於控制CE及KD loss的比例使student和teacher在學習上不因差距過大問題造成學習上的不穩定  


*   在**Response-Based KD**的整體loss function的設計中，主要是由兩個loss組成，分別是Cross entropy loss及由KL散度構成的loss，前者用於分類任務，後者用於讓student模仿teacher的輸出分佈




## 3. Feature-based KD (12%)
> Please explain the following:
> - How you extract features from the choosing intermediate layers?
> - How you design the loss?

*  關於特徵萃取，主要是分別從teacher和student模型架構的中間網路層進行特徵萃取，其中teacher也會透過中間層將這些豐富的特徵資訊傳遞給student來讓其學習有效特徵  


*   關於**Feature-based KD**的整體loss function的設計，主要由兩大部分構成，一個是cross entropy loss作為student在分類任務中，另一個loss是L2 loss作為讓student去學習teacher的特徵分佈
